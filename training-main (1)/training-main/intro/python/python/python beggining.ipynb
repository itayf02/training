Question:
Write a program that calculates and prints the value according to the given formula:
Q = Square root of [(2 * C * D)/H]
Following are the fixed values of C and H:
C is 50. H is 30.
D is the variable whose values should be input to your program in a comma-separated sequence.
Example
Let us assume the following comma separated input sequence is given to the program:
100,150,180
The output of the program should be:
18,22,24

Hints:
If the output received is in decimal form, it should be rounded off to its nearest value (for example, if the output received is 26.0, it should be printed as 26)
In case of input data being supplied to the question, it should be assumed to be a console input. 



```python
def simple_calc(C=50, H=30):
    list_of_D_temp = input().split(",")
    list_of_D = [int(k) for k in list_of_D_temp]
    for i, D in enumerate(list_of_D):
        Q = ((2 * C * D) / H)**0.5
        if i == len(list_of_D)-1:
            print(round(Q), end="")
        else:
            print(round(Q), end=",")

simple_calc()
```

Write a program which takes 2 digits, X,Y as input and generates a 2-dimensional array. The element value in the i-th row and j-th column of the array should be i*j.

Example

Question:
Define a class with a generator which can iterate the numbers, which are divisible by 7, between a given range 0 and n.

Hints:
Consider use yield



```python
class BasicGen():
    def __init__(self, integer):
        self.integer = integer
    
    def basic_iter(self):
        for i in range(self.integer):
            if i % 7 == 0:
                yield i

gen = BasicGen(30)
for num in gen.basic_iter():
    print(num)
```

Question:
A robot moves in a plane starting from the original point (0,0). The robot can move toward UP, DOWN, LEFT and RIGHT with a given steps. The trace of robot movement is shown as the following:
UP 5
DOWN 3
LEFT 3
RIGHT 2
¡­
The numbers after the direction are steps. Please write a program to compute the distance from current position after a sequence of movement and original point. If the distance is a float, then just print the nearest integer.
Example:
If the following tuples are given as input to the program:
UP 5
DOWN 3
LEFT 3
RIGHT 2
Then, the output of the program should be:
2

Hints:
In case of input data being supplied to the question, it should be assumed to be a console input.



```python
trace = input().split(" ")
x_pos, y_pos = 0, 0
for i in range(0, len(trace), 2):
    if trace[i] == "UP":
        y_pos += int(trace[i+1])
    if trace[i] == "DOWN":
        y_pos -= int(trace[i+1])
    if trace[i] == "RIGHT":
        x_pos += int(trace[i+1])
    if trace[i] == "LEFT":
        x_pos -= int(trace[i+1])
dist = round((x_pos**2 + y_pos**2)**0.5) 
print(dist)
```


Define a class named Shape and its subclass Square. The Square class has an init function which takes a length as argument. Both classes have a area function which can print the area of the shape where Shape's area is 0 by default.

Hints:
To override a method in super class, we can define a method with the same name in the super class.



```python
import abc

class Shape():
    def __init__(self):
        pass
    
    @abc.abstractmethod
    def getArea(self):
        pass

class Square(Shape):
    def __init__(self, length):
        super().__init__()
        self.length = length
        
    def getArea(self):
        return self.length**2

s = Square(2)
print(s.getArea())
```

Please write a binary search function which searches an item in a sorted list. The function should return the index of element to be searched in the list.


Hints:
Use if/elif to deal with conditions.




```python
lst = [1,4,7,10,11,18,20,22,32]
x = 15
start = 0
n = len(lst)
end = n-1
mid = n//2
while start < end:
    print(start)
    print(mid)
    print(end)
    if x == lst[mid]:
        print("found")
        break
    elif x < lst[mid]:
        end = mid
        mid = (start + end) // 2
    else:
        start = mid
        mid = (start + end) // 2
    if (end == start + 1):
        print("OOF!")
        break
```

Question: Write a Python program of recursion list sum.
Example: [1, 2, [3,4], [5,6]]

```python
def rec_sum(lst, s):
    for elem in lst:
        if type(elem) == int:
            s += elem
        elif type(elem) == list:
            print(s)
            s += rec_sum(elem, 0)
    return s
lst = [1,2,[3,4],[5,6]]
print(rec_sum(lst,0))
```
