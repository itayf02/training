    1. Write a Date class
        a. you can save only milliseconds as data member
        b. these operators must work: < > == != + - ...
        c. you can access fields using a key: date['day'], date['month']

```python
import math
SECONDS_IN_DAY: int = 86400

class Date:
    def __init__(self, mil):
        self.time = mil
    
    def __add__(self, other):
        return Date(self.time + other.time)
    
    def __eq__(self, other):
        return self.time == other.time
    
    def __lt__(self, other):
        return self.time < other.time
    
    def __gt__(self, other):
        return self.time > other.time

    def __ne__(self, other):
        return self.time != other.time
    
    def __sub__(self, other):
        return Date(self.time - other.time)
    
    def __getitem__(self, key):
        days = self.time // (1000 * SECONDS_IN_DAY)
        year = 1970
        days_in_month = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
        while True:
            days_in_year = 365
            if days < days_in_year:
                break
            days -= days_in_year
            year += 1
        if key == 'year':
            return year
        month = 1
        for days_in_this_month in days_in_month:
            if days < days_in_this_month:
                break
            days -= days_in_this_month
            month += 1
        if key == 'month':
            return month
        day = days + 1
        if key == 'day':
            return day
        if key =='hour':
            seconds = self.time // 1000
            remaining_seconds = seconds % SECONDS_IN_DAY
            hours = remaining_seconds // 3600
            return hours
```

    3. A gold mine! Your next task is to turn this data into entities using python classes.
        Hint: classes are NOT DataFrame - you need User, Message, Date ...

```python
from datetime import datetime

class DateTime:
    def __init__(self, date_str, time_str):
        self.date = datetime.strptime(date_str, "%m/%d/%y")
        self.time = datetime.strptime(time_str, "%H:%M").time()

    def __str__(self):
        return f"{self.date.strftime('%Y-%m-%d')} {self.time.strftime('%H:%M')}"
    
    def to_dict(self):
        return {"date": self.date.strftime("%Y-%m-%d"), "time": self.time.strftime("%H:%M")}

class User:
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return self.name

    def to_dict(self):
        return {"name": self.name}

class Message:
    def __init__(self, datetime_obj, user, content):
        self.datetime = datetime_obj
        self.user = user
        self.content = content

    def to_dict(self):
        return {
            "datetime": self.datetime.to_dict(),
            "user": self.user.to_dict(),
            "content": self.content,
        }

    def __str__(self):
        return f"[{self.datetime}] {self.user}: {self.content}"
```

    4. Classy. Now turn your classes into a SlowDataFrame. Oh! Not before you save them to a file first... 
        *** ALL DATA IN ONE FILE ***

```python
def parse_line(line):
    try:
        date_time_part, message_part = line.split(" - ", 1)
        date_str, time_str = date_time_part.split(", ")
        user_and_content = message_part.split(": ", 1)
        
        if len(user_and_content) == 2:
            user_name, content = user_and_content
        else:
            user_name = "System"
            content = user_and_content[0]
        
        datetime_obj = DateTime(date_str, time_str)
        user_obj = User(user_name)
        message_obj = Message(datetime_obj, user_obj, content)
        return message_obj
    except ValueError:
        return None

```

```python
import json
file_path = r"C:\Users\itayf\PycharmProjects\AI_training\training-main (1)\training-main\intro\python\python\advanced_python\resources\data\big_brother.txt"
def process_file(file_path):
    messages = []

    with open(file_path, "r", encoding="utf-8") as file:
        for line_ in file:
            line = line_.encode("ISO-8859-1").decode("utf-8")
            line = line.strip()
            if line:  # Skip empty lines
                message = parse_line(line)
                if message:
                    messages.append(message.to_dict())

    with open("messages.json", "w", encoding="utf-8") as json_file:
        json.dump(messages, json_file, indent=4)

    return messages
process_file(file_path)
```
