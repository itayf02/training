## Timing

    1. Remember slow-pandas? Can u measure how much time 'spd.do_slow_stuff' takes?
        *** FIND 4 WAYS TO FIND OUT ***

```python
import timeit
def foo(x):
    return(x**2)

#Option one:
start_time = time.time()
spd.do_slow_stuff(foo)
print(time.time() - start_time)

#option two:

start_time = time.perf_counter()
spd.do_slow_stuff(foo)
print(time.perf_counter() - start_time)

#option three:
print(timeit.timeit(lambda: do_slow_stuff(foo), number=1))
```

    2. What is faster? join strings with '+' or using '%'?

```python
str1 = " hello my"
str2 = "friend"
start_time = time.time()
_ = str1 + str2
print(time.time() - start_time)
start_time = time.time()
_ = "hello my %s" %("friend")
print(time.time() - start_time)

#the first one is faster
```

    3. Compile or to not compile - that is the question. What is faster? compile a regex before 100 iterations or not?

```python
string = "The sun rose slowly over the hills casting a warm golden glow across the valley Birds chirped cheerfully as the gentle breeze rustled the leaves Trees stood tall swaying lightly in the soft wind The air smelled fresh and clean a reminder of nature's beauty The fields stretched far dotted with colorful wildflowers Bees buzzed busily moving from bloom to bloom Life seemed peaceful and still A stream flowed quietly sparkling under the morning light In this moment everything felt simple and serene untouched by the rush of the world It was a perfect start to a new day"
```

```python
import re
import time
start_time = time.time()
# No compile
for i in range(1000):
    re.findall(r"the", string)
print(time.time() - start_time)

#With compile
compiled_regex = re.compile(r"the")
start_time = time.time()
for i in range(1000):
    compiled_regex.findall(string)
print(time.time() - start_time)
```

    2. Meet 'spd.slow_calc' - Can u run 100 of it? FAST AS POSSIBLE
        Hint: It's not as before. Consider Queue?

```python
from concurrent.futures import ProcessPoolExecutor
import random
def slow_calc():
    slow_factor = 0.1
    start_t = time.time()
    end_t = start_t + (random.random() / slow_factor)
    while time.time() < end_t:
        continue

def run_slow_calc(_):
    slow_calc()

def run_fast():
    with ProcessPoolExecutor() as executor:
        executor.map(run_slow_calc, range(100))

start_time = time.time()
run_fast()
print(time.time() - start_time)
```
